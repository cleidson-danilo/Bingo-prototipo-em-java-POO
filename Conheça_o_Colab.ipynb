```
import numpy as np
import pandas as pd

#inicializa um gerador de números aleatorios
rng= np.random.default_rng(42)

#criar um dataframe do pandas com dados simulados
dados = pd.DataFrame({
    "usuarios_afetados": rng.integers(1, 201, 600),
    "risco_segurança": rng.integers(0, 2, 600),
    "sistema_indisponivel": rng.integers(0, 2, 600),
    "impacto_financeiro": rng.integers(0, 2, 600),
})
print(dados)
```

```
#Define as regras lógica Classificar se um chamado é urgente
dados["urgente"]=(
    #regra 1
    (dados["risco_segurança"]==1)
    #regra 2
    |((dados["sistema_indisponivel"]==1)
    &(dados["usuarios_afetados"]>40))
    #regra 3
    |((dados["impacto_financeiro"]==1)
    &(dados["usuario_afetados"]>20))
).astype(int)#converte o resultado booleano (True/false) para inteiro

print (dados)
```
